```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
```

```python
import pandas as pd

dados = pd.read_csv("dataset_features.csv")

dados.head()
```

```python
dados.info()
```

```python
dados.describe()
```

```python
dados.isnull().sum()
```

```python
plt.figure(figsize=(8,5))

dados["valor_total"].hist(bins=10)

plt.title("Distribuição do Valor Total das Vendas")
plt.xlabel("Valor Total")
plt.ylabel("Frequência")

plt.show()
```

```python
dados.groupby("produto")["valor_total"].sum().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Total de Vendas por Produto")
plt.ylabel("Valor Total")

plt.show()
```

```python
plt.figure(figsize=(8,5))

dados["idade"].hist(bins=8)

plt.title("Distribuição das Idades")
plt.xlabel("Idade")
plt.ylabel("Frequência")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=dados["valor_total"])

plt.title("Boxplot do Valor Total")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    dados.corr(numeric_only=True),
    annot=True,
    cmap="Blues"
)

plt.title("Matriz de Correlação")

plt.show()
```
